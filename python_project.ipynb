# Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Importing Dataset

```python
dataset = pd.read_csv('heart.csv')
X= dataset.iloc[:,:-1].values
Y= dataset.iloc[:,-1].values
```

```python
print(X)
```

```python
print(Y)
```

# Data Preprocessing

```python
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
le2 = LabelEncoder()
le6 = LabelEncoder()
le8 = LabelEncoder()
le10 = LabelEncoder()
X[:,1] = le1.fit_transform(X[:,1])
X[:,2] = le2.fit_transform(X[:,2])
X[:,6] = le6.fit_transform(X[:,6])
X[:,8] = le8.fit_transform(X[:,8])
X[:,10] = le10.fit_transform(X[:,10])
```

```python
print(X)
```

# Splitting Dataset into Training and Test set

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
print(X_train)
```

```python
print(Y_test)
```

# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.fit_transform(X_test)
import joblib
joblib.dump(sc, 'scaler.pkl')
```

```python
print(X_train)
```

# Training Dataset

```python
#SVM
from sklearn.svm import SVC
model_svm = SVC()
model_svm.fit(X_train,Y_train)
```

```python
import joblib
joblib.dump(model_svm, 'heart_model.pkl')
print("Model saved as heart_model.pkl")
```

```python
#Decision Tree
from sklearn.tree import DecisionTreeClassifier
model_tree = DecisionTreeClassifier()
model_tree.fit(X_train,Y_train)
```

```python
#Random Forest
from sklearn.ensemble import RandomForestClassifier
model_random = RandomForestClassifier()
model_random.fit(X_train,Y_train)
```

```python
y_pred_svm = model_svm.predict(X_test)
y_pred_tree = model_tree.predict(X_test)
y_pred_random = model_random.predict(X_test)
```

# Testing Accuracy on different ML Algorithms

```python
from sklearn.metrics import accuracy_score
SVM_Acc = accuracy_score(Y_test,y_pred_svm)
DecTree_Acc = accuracy_score(Y_test,y_pred_tree)
RanFor_Acc = accuracy_score(Y_test,y_pred_random)
```

```python
print(SVM_Acc)
print(DecTree_Acc)
print(RanFor_Acc)
```

# Plotting Accuracy of different ML Algorithms

```python
plt.bar("Support Vector Machine",SVM_Acc,width=0.5)
plt.bar("Decision Tree",DecTree_Acc,width=0.5)
plt.bar("Random Forest",RanFor_Acc,width=0.5)
plt.xlabel('Machine Learning Algorithms')
plt.ylabel('Accuracy')
```
